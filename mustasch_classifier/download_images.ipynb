# Ladda ner träningsbilder
Laddar ner bilder från Google Images för epic och thin-mapparna.

```python
from icrawler.builtin import BingImageCrawler

queries_epic = [
    "thick horseshoe mustache man portrait",
    "tom selleck mustache",
    "ron swanson mustache",
    "freddie mercury mustache",
    "handlebar mustache man portrait",
    "bushy mustache close up",
    "walrus mustache man",
]

queries_thin = [
    "thin pencil mustache man portrait",
    "small mustache man close up",
    "light stubble mustache portrait",
    "weak mustache man",
]


for query in queries_epic:
    print(f"Laddar: {query}")
    crawler = BingImageCrawler(
        storage={"root_dir": "data/epic_dataset/epic"},
        log_level=50
    )
    crawler.crawl(keyword=query, max_num=20)

for query in queries_thin:
    print(f"Laddar: {query}")
    crawler = BingImageCrawler(
        storage={"root_dir": "data/epic_dataset/thin"},
        log_level=50
    )
    crawler.crawl(keyword=query, max_num=20)

print("Klart!")
```

## Kolla hur många bilder vi har nu

## Rensa datan manuellt
Gå igenom mapparna och ta bort:
- Bilder utan ansikte
- Tecknade bilder / illustrationer
- Bilder där mustaschen inte syns tydligt
- Felaktiga träffar

Ta sedan bort korrupta bildfiler:

```python
from PIL import Image
import os

removed = 0

for folder in ["data/epic_dataset/epic", "data/epic_dataset/thin"]:
    for fname in os.listdir(folder):
        fpath = os.path.join(folder, fname)
        try:
            img = Image.open(fpath)
            img.verify()
        except Exception:
            print(f"Tar bort korrupt fil: {fpath}")
            os.remove(fpath)
            removed += 1

print(f"Tog bort {removed} korrupta filer")
```
