```python
def extract_tags(x):

    if isinstance(x, dict):
        return list(x.keys())

    elif isinstance(x, list):
        return x

    else:
        return []
```

```python
df_model["tag_names"] = (
    df_model["tags"]
    .apply(extract_tags)
)
```

```python
df_model["tag_names"].head()
```

```python
df_model["tag_names"].info()
```

```python
from collections import Counter

tag_counter = Counter()

for tags in df_model["tag_names"]:
    tag_counter.update(tags)

tag_counter.most_common(40)
```

```python
top_tags = [tag for tag, count in tag_counter.most_common(50)]

def keep_top_tags(tags):
    return [tag for tag in tags if tag in top_tags]

df_model["top_tags"] = df_model["tag_names"].apply(keep_top_tags)
```

Samla top 50 tags och binärisera dem som genres och kategorier

```python
from sklearn.preprocessing import MultiLabelBinarizer

mlb_tags = MultiLabelBinarizer()

tags_encoded = pd.DataFrame(
    mlb_tags.fit_transform(df_model["top_tags"]),
    columns=[f"tag_{c}" for c in mlb_tags.classes_],
    index=df_model.index
)
```

```python
df_final_tags = pd.concat(
    [
        df_base,
        genre_encoded,
        categories_encoded,
        tags_encoded
    ],
    axis=1
)
```

Modellen var något bättre på att identifiera spel med svagare användarreception än att identifiera högpresterande titlar, vilket kan tyda på att negativa mönster är mer konsekventa än framgångsmönster.

```python
from sklearn.metrics import confusion_matrix
import seaborn as sns
import matplotlib.pyplot as plt

cm = confusion_matrix(y_test, y_pred)

plt.figure(figsize=(6,5))

sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    cmap="Blues",
    xticklabels=["Not Successful", "Successful"],
    yticklabels=["Not Successful", "Successful"]
)

plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.title("XGBoost Confusion Matrix - Test Data")

plt.show()
```

```python
from sklearn.metrics import confusion_matrix
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np

cm = confusion_matrix(y_test, y_pred)

# Gör om till procent per rad
cm_percent = cm.astype("float") / cm.sum(axis=1)[:, np.newaxis]

plt.figure(figsize=(6,5))

sns.heatmap(
    cm_percent,
    annot=True,
    fmt=".2%",
    cmap="Blues",
    xticklabels=["Not Successful", "Successful"],
    yticklabels=["Not Successful", "Successful"]
)

plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.title("XGBoost Confusion Matrix (%)")

plt.show()
```

Spelreception verkar delvis förutsägbar genom gameplay-, community- och Steam-relaterade features, men resultaten antyder också att en stor del av spelens framgång beror på faktorer som inte fångas i metadata ensam.

Modellen hade något lättare att identifiera “Not Successful”-spel än “Successful”-spel. Det kan delvis bero på att negativa mönster är mer konsekventa, men också på att datan innehöll fler mindre framgångsrika spel än framgångsrika spel.
