```python
def extract_tags(x):

    if isinstance(x, dict):
        return list(x.keys())

    elif isinstance(x, list):
        return x

    else:
        return []
```

```python
df_model["tag_names"] = (
    df_model["tags"]
    .apply(extract_tags)
)
```

```python
df_model["tag_names"].head()
```

```python
df_model["tag_names"].info()
```

```python
from collections import Counter

tag_counter = Counter()

for tags in df_model["tag_names"]:
    tag_counter.update(tags)

tag_counter.most_common(40)
```

```python
top_tags = [tag for tag, count in tag_counter.most_common(50)]

def keep_top_tags(tags):
    return [tag for tag in tags if tag in top_tags]

df_model["top_tags"] = df_model["tag_names"].apply(keep_top_tags)
```

Testa PCA först

```python
# PCA-experiment: copy av X/y
X_pca_source = X.copy()
y_pca = y.copy()
```

```python
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X_pca_source)

pca = PCA(n_components=0.95)
X_pca = pca.fit_transform(X_scaled)

print("Original features:", X_pca_source.shape)
print("PCA features:", X_pca.shape)
```

```python
from sklearn.model_selection import train_test_split

X_train_pca, X_test_pca, y_train_pca, y_test_pca = train_test_split(
    X_pca,
    y_pca,
    test_size=0.2,
    random_state=42
)
```

```python
from xgboost import XGBClassifier

xgb_pca = XGBClassifier(
    subsample=0.8,
    n_estimators=200,
    min_child_weight=3,
    max_depth=6,
    learning_rate=0.05,
    colsample_bytree=1.0,
    eval_metric="logloss",
    random_state=42
)

xgb_pca.fit(X_train_pca, y_train_pca)
```

```python
from sklearn.metrics import classification_report

y_pred_pca = xgb_pca.predict(X_test_pca)

print(classification_report(y_test_pca, y_pred_pca))
```

```python
import matplotlib.pyplot as plt
import numpy as np

explained = np.cumsum(
    pca.explained_variance_ratio_
)

plt.figure(figsize=(8,5))

plt.plot(
    explained,
    marker="o"
)

plt.xlabel("Number of Components")
plt.ylabel("Cumulative Explained Variance")

plt.title("PCA Explained Variance")

plt.grid()

plt.show()
```

Variansen i datan var spridd över många dimensioner snarare än koncentrerad till några få starka komponenter. Detta kan förklara varför PCA inte förbättrade modellen, samtidigt som tree-baserade modeller som Random Forest och XGBoost fortfarande kunde hitta användbara lokala mönster i datan.

PCA reducerade antalet dimensioner men förbättrade inte modellen. För tree-baserade modeller verkade de ursprungliga featuresen innehålla mer användbar signal än de transformerade huvudkomponenterna.

```python
from xgboost import XGBRegressor

xgb_reg = XGBRegressor(
    subsample=0.8,
    n_estimators=200,
    min_child_weight=3,
    max_depth=6,
    learning_rate=0.05,
    colsample_bytree=1.0,
    random_state=42
)

xgb_reg.fit(X_train_reg, y_train_reg)
```
